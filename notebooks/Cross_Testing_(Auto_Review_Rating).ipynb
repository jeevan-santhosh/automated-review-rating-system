# **IMPORT NECESSARY LIBRARIES**

```python
import pickle
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# **LOAD MODELS AND VECTORIZERS**

```python
try:
    with open('Model_A.pkl', 'rb') as f:
        model_A = pickle.load(f)
    print("Model_A (Trained on Balanced Data) loaded successfully.")

    with open('Model_B.pkl', 'rb') as f:
        model_B = pickle.load(f)
    print("Model_B (Trained on Imbalanced Data) loaded successfully.")

    with open('balanced_tfidf_vectorizer.pkl', 'rb') as f:
        vectorizer_balanced = pickle.load(f)
    print("Balanced TF-IDF Vectorizer loaded successfully.")

    with open('imbalanced_tfidf_vectorizer.pkl', 'rb') as f:
        vectorizer_imbalanced = pickle.load(f)
    print("Imbalanced TF-IDF Vectorizer loaded successfully.")

except FileNotFoundError as e:
    print(f"Error: {e}. Please ensure you have uploaded all four files to your Colab session.")
```

# **LOAD TEST DATA**

```python
imbalanced_df = pd.read_csv('imbalanced_test_final.csv')
balanced_df = pd.read_csv('balanced_test_final.csv')
```

```python
# Split into features and labels
X_test_imbal = imbalanced_df['Review']
y_test_imbal = imbalanced_df['Rating']

X_test_bal = balanced_df['Review']
y_test_bal = balanced_df['Rating']
```

# **EVALUATION FUNCTION**

```python
# Transform text data using respective vectorizers
X_test_bal_vect = vectorizer_balanced.transform(X_test_bal)
X_test_imbal_vect = vectorizer_imbalanced.transform(X_test_imbal)
```

```python
# Make predictions
y_pred_a_bal = model_A.predict(X_test_bal_vect)
y_pred_b_imbal = model_B.predict(X_test_imbal_vect)
```

```python
# Transform text data using respective vectorizers
X_test_bal_vect_a = vectorizer_balanced.transform(X_test_bal)
X_test_imbal_vect_b = vectorizer_imbalanced.transform(X_test_imbal)
```

```python
# Cross-transform for testing on opposite datasets
X_test_bal_vect_b = vectorizer_imbalanced.transform(X_test_bal)
X_test_imbal_vect_a = vectorizer_balanced.transform(X_test_imbal)
```

```python
# Make predictions (Cross-testing and Control-testing)
y_pred_a_imbal = model_A.predict(X_test_imbal_vect_a) # Model A on Imbalanced Test Set
y_pred_b_bal = model_B.predict(X_test_bal_vect_b)   # Model B on Balanced Test Set
```

# **CROSS TESTING**

```python
# Evaluate Models

def evaluate_model(y_true, y_pred, model_name):
    print(f"===== Evaluation for {model_name} =====")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("\nClassification Report:\n", classification_report(y_true, y_pred, digits=4))
    print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred))
    print("\n--------------------------------------\n")

# Cross Tests
evaluate_model(y_test_imbal, y_pred_a_imbal, "Model_A on Imbalanced Test Set")
evaluate_model(y_test_bal, y_pred_b_bal, "Model_B on Balanced Test Set")

# Control Tests
evaluate_model(y_test_bal, y_pred_a_bal, "Model_A on Balanced Test Set")
evaluate_model(y_test_imbal, y_pred_b_imbal, "Model_B on Imbalanced Test Set")
```
