### **LOAD DATASETS**

```python
df1 = pd.read_csv('/content/flipkart_main.csv', encoding='latin-1')
df2 = pd.read_csv('/content/flipkart_product_main.csv', encoding='latin-1')
df3 = pd.read_csv('/content/flipkart_reviews_dataset_main.csv', encoding='latin-1')
```

### **COMBINING DATASETS**

```python
expanded_dataset = pd.concat([df1, df2, df3], ignore_index=True)
expanded_dataset.to_csv("expanded_dataset_final.csv", index=False)
print("Expanded dataset saved: expanded_dataset_final.csv")
print(f"Dataset shape: {expanded_dataset.shape}")
expanded_dataset.head()
```

```python
expanded_dataset = expanded_dataset.dropna(subset=['Review'])
columns_to_keep = ['Review', 'Rating']
dataset = expanded_dataset[columns_to_keep].copy()
expanded_dataset.head()
```

```python
# Bar Plot
plt.figure(figsize=(8, 6))
sns.countplot(x='Rating', data=expanded_dataset, palette="viridis", edgecolor='black', zorder=3)
plt.title('Bar Plot of Expanded Dataset', fontsize=14)
plt.xlabel('Review Rating', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.show()
```
